```python
#3-create the same customer satisfaction data using a python dictionary and convert it into a pandas series
cust_data=pd.Series({'C001':8,'C002':6,'C003':9,'C004':4,'C005':7,'C006':10,'C007':5,'C008':8})
print(cust_data)
```
