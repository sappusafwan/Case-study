```python
import numpy as np
```

```python
emp=np.array(['arun','bhavana','charan','divya','eshan','fathima'])
print(emp)

performance = np.array([
    [78, 82, 85, 88],
    [65, 70, 68, 74],
    [92, 89, 95, 91],
    [55, 62, 60, 67],
    [81, 79, 84, 87],
    [72, 75, 78, 80]
])
print(performance)

quarters = np.array(["Q1", "Q2", "Q3", "Q4"])
print(quarters)
```

```python
#1.Calculate the total performance score for every employee.
total_performance_score=performance.sum(axis=1)
print(f'total performance score of every employee is:{total_performance_score}')
```

```python
#2. Calculate the average quarterly performance for every employee.
average_quarterly_performance=performance.mean(axis=1)
print(f'avg quarterly performance of every employee is:{average_quarterly_performance}')
```

```python
#3. Calculate the average performance for each quarter.
average_performance=performance.mean(axis=0)
print(f'avg performance for each quarter is:{average_performance}')
```

```python
#4. Identify the employee with the highest total performance score.
total=performance.sum(axis=1)
score=np.argmax(total)
print(f'employee with highest total score is:{emp[score]} with {total[score]}')
```

```python
#5. Rank all employees based on their total performance score
total=performance.sum(axis=1)
rank=np.argsort(total)[::-1]
print(rank)

for i in rank:
    print(emp[i],performance[i])
```

```python
#6.Calculate the quarter-over-quarter percentage growth for every employee from q1 to q4
print(performance[:,1:])
```

```python
per_growth=(performance[:,1:]-performance[:,:-1])/performance[:,:-1]*100
print(per_growth)
```

```python
#7.identify employees whose average performance is gerater than or equal to 80
avg=performance.mean(axis=1)
p=(emp[avg>=80])
print(p)
```

```python
#8.count how many employees have an average performance below 70
avg_count=np.sum(avg<70)
print(avg_count)
```

```python
#9.find the highest performance score recorded in the entire dataset
highest=performance.max()
print(highest)
```

```python
#10.identify the employee and quarter in which the highest performance score occured
highest_performance=performance.max()
score=np.where(performance==highest_performance)
print(emp[score[0][0]])
print(quarters[score[1][0]])
print(highest_performance)
```
