```python
#13. Write a function grade(score) that returns:

def grade(scores):
    if scores>=90:
        return('Distinction')
    elif scores>=75 & scores<=89:
        return('First Class')
    elif scores>=60 & scores<=74:
        return('Second Class')
    else:
        return('Fail')
print(scores.apply(grade))
```

```python
#14. Display how many students fall in each grade, then the same as percentages.
print(f'the students in each grade is:{scores.value_counts()}')
print(f'the percentage of students in each grade is:{(scores.value_counts(normalize=True)*100)}')
```

```python
#15. Display the unique grades and the number of unique grades.
print(f'the unique grades are:{scores.unique()}')
print(f'the no of unique grades are:{scores.nunique()}')
```
